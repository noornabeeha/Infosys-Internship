```python
# 2. Build Threat Scoring Logic
#score = 5
# credential	+3
# execution	+2
# privilege	+3
# persistence	+2
# lateral	+2
# print(data['objects'][0]['name'])
# print(data['objects'][1]['name'])
print('--------------------------------')
print('---------function starts----------')
# for Keyword (in name)	Score Increase, name is the key present in 'objects' dict
def threat_score(tech):
    score = 5
    if "credential" in tech:
        score+=3
    if "execution" in tech:
        score+=2
    if "privilege" in tech:
        score+=3
    if "persistence" in tech:
        score+=2
    if "lateral" in tech:
        score+=2
    return score
# print(threat_score(data['objects'][0]['name']))
# print(threat_score(data['objects'][1]['name']))
# print(threat_score(data['objects'][100]['name']))
for i in range(800):
    riskname = (threat_score(data['objects'][i]['name']))
    print(data['objects'][i]['name'],"--->",riskname)
```

```python
# 'description'

def threat_score_desc(desc):
    score = 5
    if "administrator" in desc:
        score+=2
    if "remote" in desc:
        score+=2
    if "bypass" in desc:
        score+=2
    if "stealth" in desc:
        score+=1
    return score
for i in range(100):
    des = (threat_score_desc(data['objects'][i]['description']))
    print(data['objects'][i]['name'],"--->",des)
```
